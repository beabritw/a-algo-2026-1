```python
import heapq

# N pacientes com níveis de dor (1-10)
pacientes = [
    ("Maria", 3),
    ("João", 9),
    ("Ana", 5),
    ("Pedro", 10),
    ("Carlos", 7)
]

# Inicializa o Max-Heap
max_heap = []

# Insere os pacientes invertendo o valor da dor (p/ prioridade)
for nome, dor in pacientes:
    # O heapq ordena pelo menor número, entao -10 vem antes de -9, -7, etc.
    heapq.heappush(max_heap, (-dor, nome))

# Processa os pacientes do mais grave para o menos grave
while max_heap:
    dor_negativa, nome = heapq.heappop(max_heap)
    dor_real = -dor_negativa  # valor real da dor
    print(f"Paciente: {nome} | Nível de dor: {dor_real}")
```
